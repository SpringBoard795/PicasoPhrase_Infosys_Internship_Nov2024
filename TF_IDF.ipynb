```python
from sklearn.feature_extraction.text import TfidfVectorizer
import heapq
import pandas as pd
```

```python
def find_top_n_words_by_tfidf(sentences, n=10):
  tfidf = TfidfVectorizer()
  X = tfidf.fit_transform(sentences)
  feature_names = tfidf.get_feature_names_out()
  df_tfidf = pd.DataFrame(X.toarray(), columns=feature_names)
  top_n_words = []
  for column in df_tfidf.columns:
    top_n_words.append((column, df_tfidf[column].max()))

  top_n_words = heapq.nlargest(n, top_n_words, key=lambda x: x[1])
  return top_n_words
```

```python
sentences = [
    "Biographical works are usually non-fiction, but fiction can also be used to portray a person's life.",
    "One in-depth form of biographical coverage is called legacy writing.",
    "Works in diverse media, from literature to film, form the genre known as biography.",
    "An authorized biography is written with the permission, cooperation, and at times, participation of a subject or a subject's heirs."
]
```

```python
top_words = find_top_n_words_by_tfidf(sentences, n=7)
top_words
```
