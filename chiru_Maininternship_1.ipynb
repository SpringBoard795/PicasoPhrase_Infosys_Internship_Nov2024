```python
import pandas as pd
from scipy import stats
import re
import numpy as np
def extract_numeric_values(caption):
    age_match = re.search(r'(\d+)\s*year', caption)
    score_match = re.search(r'score\s*of\s*([\d.]+)(?!\.)', caption)

    age = int(age_match.group(1)) if age_match else None
    try:
        score = float(score_match.group(1)) if score_match else None
    except ValueError:
        score = None

    return pd.Series({'age': age, 'score': score})

captions_df = pd.read_csv("/content/all_data/test/radiology/captions.txt", sep="\t", header=None, names=['image_id', 'caption'])
numeric_data = captions_df['caption'].apply(extract_numeric_values)
mean_values = numeric_data.mean()
median_values = numeric_data.median()
mode_values = numeric_data.apply(lambda x: stats.mode(x)[0][0] if isinstance(stats.mode(x)[0], np.ndarray) and len(stats.mode(x)[0]) > 0 else (stats.mode(x)[0] if not np.isnan(stats.mode(x)[0]) else np.nan))

print("Mean:", mean_values)
print("Median:", median_values)
print("Mode:", mode_values)
```

```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec

def grayscale_to_color(image_path):
    gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if gray_image is None:
        print(f"Error: Unable to load image at path {image_path}")
        return None
    color_image = cv2.applyColorMap(gray_image, cv2.COLORMAP_JET)
    color_image_rgb = cv2.cvtColor(color_image, cv2.COLOR_BGR2RGB)
    return color_image_rgb

image_paths = [
    '/content/all_data/test/radiology/images/PMC107839_1471-2296-3-6-3.jpg'
]

captions = [os.path.basename(path) for path in image_paths]
images = []
valid_captions = []
for i, path in enumerate(image_paths):
    image = grayscale_to_color(path)
    if image is not None:
        images.append(image)
        valid_captions.append(captions[i])
if not images:
    print("No valid images to display.")
else:
    fig = plt.figure(figsize=(10, 8))
    grid = GridSpec(len(images), 2, width_ratios=[3, 1])

    for i in range(len(images)):
        ax_img = fig.add_subplot(grid[i, 0])
        ax_img.imshow(images[i])
        ax_img.axis('off')

        ax_caption = fig.add_subplot(grid[i, 1])
        ax_caption.text(0.5, 0.5, valid_captions[i], va='center', ha='left', wrap=True, fontsize=10)
        ax_caption.axis('off')

    plt.tight_layout()
    plt.show()
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import numpy as np
import matplotlib.pyplot as plt
import cv2
def grayscale_to_color(image_path):
    """
    Converts a grayscale image to a color image using a colormap.

    Args:
        image_path: Path to the grayscale image file.

    Returns:
        A NumPy array representing the color image (RGB format).
    """

    gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
    if gray_image is None:
        print(f"Error: Unable to load image at path {image_path}")
        return None
    color_image = cv2.applyColorMap(gray_image, cv2.COLORMAP_JET)
    color_image_rgb = cv2.cvtColor(color_image, cv2.COLOR_BGR2RGB)
    return color_image_rgb

image_path = '/content/all_data/validation/radiology/images/PMC1325037_1742-6413-2-21-1.jpg'
color_image = grayscale_to_color(image_path)

datagen = ImageDataGenerator(
    rotation_range=20,
    width_shift_range=0.1,
    height_shift_range=0.1,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)
color_image = color_image.reshape((1,) + color_image.shape)

plt.figure(figsize=(10, 5))
num_augmented_images = 5
for i in range(num_augmented_images):
    augmented_image = next(datagen.flow(color_image, batch_size=1))[0]

    plt.subplot(1, num_augmented_images, i + 1)
    plt.imshow(augmented_image.astype(np.uint8))
    plt.title(f"Augmented Image {i + 1}")
    plt.axis("off")

plt.show()
```

```python
import os
import cv2
import numpy as np
from PIL import Image
data = "/content/all_data/test/radiology/images/PMC107839_1471-2296-3-6-3.jpg"

def load_and_preprocess_image(image_path, target_size=(224, 224)):
    """Loads and preprocesses a single image.

    Args:
        image_path: Path to the image file.
        target_size: Tuple (width, height) for resizing the image.

    Returns:
        A NumPy array representing the preprocessed image.
    """
    try:
        img = Image.open(image_path).convert('RGB')
        img = img.resize(target_size)
        img_array = np.array(img) / 255.0
    except Exception as e:
        print(f"Error loading image {image_path}: {e}")
        return None

    return img_array

# Load and preprocess the image
image = load_and_preprocess_image(data)

# Check if the image was loaded successfully
if image is not None:
    print("Image loaded and preprocessed successfully.")

else:
    print("Failed to load image.")
```

```python
import pandas as pd
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, RegexpStemmer, WordNetLemmatizer
from nltk import pos_tag, ne_chunk
from sklearn.feature_extraction.text import CountVectorizer
from textblob import TextBlob
import nltk
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('wordnet')
dataset_path = '/content/all_data/train/radiology/captions.txt'
data = pd.read_csv(dataset_path, delimiter='\t', on_bad_lines='warn')
print(data.columns)
caption_column_name = data.columns[1]
stop_words = set(stopwords.words('english'))
captions = data[caption_column_name][:5]
```

Stemming

```python
#Stemming
porter_stemmer = PorterStemmer()
print("Porter Stemming Results:")
for caption in captions:
    stemmed_words = [porter_stemmer.stem(word) for word in word_tokenize(caption) if word.lower() not in stop_words]
    print("Original:", caption)
    print("Stemmed:", " ".join(stemmed_words))
    print("="*50)
```

Lemmatization

```python
#Lemmatization
lemmatizer = WordNetLemmatizer()
print("Lemmatization Results:")
for caption in captions:
    lemmatized_words = [lemmatizer.lemmatize(word) for word in word_tokenize(caption) if word.lower() not in stop_words]
    print("Original:", caption)
    print("Lemmatized:", " ".join(lemmatized_words))
    print("="*50)
```

RegExp Stemmer

```python
#Reg-Exp Stemmer
regexp_stemmer = RegexpStemmer('ing$|s$|ed$', min=4)
print("Regexp Stemming Results:")
for caption in captions:
    regexp_stemmed_words = [regexp_stemmer.stem(word) for word in word_tokenize(caption) if word.lower() not in stop_words]
    print("Original:", caption)
    print("Regexp Stemmed:", " ".join(regexp_stemmed_words))
    print("="*50)
```

SnowBall Stemmer

```python
#Snowball Stemmer
snowball_stemmer = SnowballStemmer("english")

print("Snowball Stemming Results:")
for caption in captions:
    snowball_stemmed_words = [snowball_stemmer.stem(word) for word in word_tokenize(caption) if word.lower() not in stop_words]
    print("Original:", caption)
    print("Snowball Stemmed:", " ".join(snowball_stemmed_words))
    print("="*50)
```

POS Tagging

```python
#POS Tagging
print("POS Tagging Results:")
for caption in captions:
    words = word_tokenize(caption)
    pos_tags = pos_tag(words)
    print("Original:", caption)
    print("POS Tags:", pos_tags)
    print("="*50)
```

NER-Name Entity Recognition

```python
#NER
print("Named Entity Recognition (NER) Results:")
for caption in captions:
    words = word_tokenize(caption)
    pos_tags = pos_tag(words)
    ner_tree = ne_chunk(pos_tags)
    print("Original:", caption)
    print("Named Entities:", ner_tree)
    print("="*50)
```

```python
import pandas as pd
import numpy as np
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

# Step 1: Load data
dataset_path = '/content/all_data/train/non-radiology/captions.txt'
data = pd.read_csv(dataset_path, sep='\t', header=None, names=['caption', 'label', 'split'])

captions = data['caption']
labels = np.random.randint(0, 2, len(captions))

# Step 2: Tokenize the captions
tokenizer = Tokenizer(num_words=5000)
tokenizer.fit_on_texts(captions)
sequences = tokenizer.texts_to_sequences(captions)

# Step 3: Pad the sequences
padded_sequences = pad_sequences(sequences, maxlen=20)

# Step 4: Split data into training and test sets
train_size = int(0.8 * len(padded_sequences))
X_train, X_test = padded_sequences[:train_size], padded_sequences[train_size:]
y_train, y_test = labels[:train_size], labels[train_size:]

# Step 5: Build a SimpleRNN-based model
model = Sequential([
    Embedding(input_dim=5000, output_dim=64, input_length=20),
    SimpleRNN(64, return_sequences=False),
    Dense(1, activation="sigmoid")
])

# Step 6: Compile the model
model.compile(optimizer="adam", loss="binary_crossentropy", metrics=["accuracy"])

# Step 7: Train the model
model.fit(X_train, y_train, epochs=10, batch_size=32, validation_split=0.2)

# Step 8: Make predictions
predictions = model.predict(X_test)

for i, prediction in enumerate(predictions[:10]):
    print(f"Caption: {data['caption'].iloc[train_size + i]}")
    print(f"Predicted Label: {round(prediction[0])}")
    print("-" * 50)
```
